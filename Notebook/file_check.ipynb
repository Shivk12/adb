```python
def file_check(file_path):
    try:
        list_file=dbutils.fs.ls(file_path)
        #print(list_file)
        if len(list_file)>0:
            file_pattern=r"loan_default_(\d{8})\.csv"
            print(file_pattern)
            for file in list_file:
                print(file.name)
                f_name=re.match(file_pattern,file.name)
                print(f_name)
                date_str=f_name.group(1)
                print(date_str)
                try:
                    valid_date = datetime.strptime(date_str,"%Y%m%d")
                    print(valid_date)
                except:
                    print("Date not proper")
                    raise RuntimeError("Date not proper")
                if re.match(file_pattern,file.name):
                    print("file is present")
                    return "file is present"
                else:
                    print("file not proper")
                    raise RuntimeError("File not proper")
    except:
        raise RuntimeError("File not present")
```
